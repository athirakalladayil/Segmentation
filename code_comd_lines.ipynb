```python
# To create the zip file
import zipfile
import os

def zip_jpg_images(folder_path):
    zip_filename = os.path.join(folder_path, 'out_rigid_hem2.zip')
    
    with zipfile.ZipFile(zip_filename, 'w') as zipf:
        for root, dirs, files in os.walk(folder_path):
            for file in files:
                if file.lower().endswith('.jpg') or file.lower().endswith('.jpeg'):
                    file_path = os.path.join(root, file)
                    zipf.write(file_path, os.path.relpath(file_path, folder_path))
                    
    print(f'Zip file created: {zip_filename}')

# Example usage
folder_path = '/home/projects/registration/athira/mytest/aug/aug15_Scale_issue_roi/out_rigid_hem2'
zip_jpg_images(folder_path)
```

# # Copy the brain datasets from one folder to another

```python
import os
import shutil # used for copying, moving, and removing files 

source_folder = '/home/projects/analytics/maadhu/fetus_brain/bfi/B_226_FB_77_BFI_output/'
destination_folder = '/home/projects/registration/athira/Multi_Modal_Registration/Data/bfi/B_226_FB_xx_BFI_output'

# Create folder if not formed
os.makedirs(destination_folder, exist_ok=True)

# Loop through all the files
for filename in os.listdir(source_folder):
    if filename.endswith("_original.jpg"):
        src_path = os.path.join(source_folder, filename)
        dst_path = os.path.join(destination_folder, filename)
        shutil.copy(src_path, dst_path)
        #print(f"copied:{filename}")
print(f"Finished copying files.")
        
```

# To delete all images inside a folder while keeping any subfolders intact

```python
import os
main_folder = '/home/'

# Loop through the items in the folder
for item in os.listdir(main_folder):
    item_path = os.path.join(main_folder, item)
    
    # Check if it's a file and not a directory
    if os.path.isfile(item_path):
        
        # Delete the file
        os.remove(item_path)
        #print(f"Deleted: {item}")

print("Finished deleting all files.")
```
